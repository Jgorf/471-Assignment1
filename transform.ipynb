# Data Transformation

## Remove Unwanted Columns

```python
data = data.drop(columns=['Block', 'IUCR', 'Description', 'Domestic', 'Beat', 'District', 'Ward', 'Community Area', 'FBI Code', 'X Coordinate', 'Y Coordinate', 'Updated On', 'Latitude', 'Longitude', 'Location'])

data
```

## Rename Column

```python
data.rename(columns={'Location Description': 'Location', 'Primary Type': 'Crime Type'}, inplace=True)
data
```

## Split Date Column

```python
# Format: month/day/year hour:minute:second AM/PM (e.g., 10/06/2026 12:00:00 AM)
data['Date'] = pd.to_datetime(data['Date'], format='%m/%d/%Y %I:%M:%S %p', errors='coerce')

data['Month'] = data['Date'].dt.month
data['Day'] = data['Date'].dt.day
data['Year'] = data['Date'].dt.year

data['Time'] = data['Date'].dt.strftime('%I:%M %p') # 12 hour clock

data[['Month', 'Day', 'Year', 'Time']]
```

## Arrest Rate

```python
data['Arrest Rate'] = data.groupby('Crime Type')['Arrest'].transform('mean')*100 # arrest rate (percentage)

data[['Crime Type', 'Arrest Rate']].drop_duplicates(subset='Crime Type').sort_values('Arrest Rate', ascending=False) # arrest rates by crime descending order
```

## Reordering Columns

```python
# Move Year, Month, and Day columns after Date, then delete Date
cols = data.columns.tolist()
for col in ['Year', 'Month', 'Day', 'Time']:
    cols.remove(col)
    
date_idx = cols.index('Date')

cols.insert(date_idx + 1, 'Year')
cols.insert(date_idx + 2, 'Month')
cols.insert(date_idx + 3, 'Day')
cols.insert(date_idx + 4, 'Time')


# remove Date column from displayed columns
cols.remove('Date')
data = data[cols]

# Move Arrest Rate after Arrest
cols = data.columns.tolist()
arrest_idx = cols.index('Arrest')
cols.remove('Arrest Rate')
cols.insert(arrest_idx + 1, 'Arrest Rate')

data
```

# Export Transformed Dataset

```python
data.to_csv('./Crime_2020-2026.csv', index=False)
```
